## Differ HDF files using h5diff 🔎


#### Imports

```python
import numpy as np
import h5py
import subprocess
```

#### Creating two similar HDF Files

```python
data = np.random.rand(5, 5)

with h5py.File('file1.h5', 'w') as f1:
    f1.create_dataset('matrix', data=data)

with h5py.File('file2.h5', 'w') as f2:
    f2.create_dataset('matrix', data=data)

with h5py.File('file1.h5', 'r') as f1, h5py.File('file2.h5', 'r') as f2:
    print("Matrix from file1:")
    print(f1['matrix'][:])
    print("\nMatrix from file2:")
    print(f2['matrix'][:])
```

#### Compare files using h5diff

```python
try:
    result = subprocess.run(['h5diff', 'file1.h5', 'file2.h5'])
    print("Files are identical" if result.returncode == 0 else "Files are different")
except Exception as e:
    print(f"Error: {e}")
```
